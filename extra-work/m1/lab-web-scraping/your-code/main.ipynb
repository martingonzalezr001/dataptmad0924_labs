#### Display the trending Python repositories in GitHub.

The steps to solve this problem is similar to the previous one except that you need to find out the repository names instead of developer names.

````python
# This is the url you will scrape in this exercise
url = 'https://github.com/trending/python?since=daily'
````

````python
# your code here
response = requests.get(url)
html = response.content
parsed_html = bs4.BeautifulSoup(html, "html.parser") # We parsed the content 
print(parsed_html) 
````

````python
my_articles =  parsed_html.find_all("article",{"class":"Box-row"})
print(len(my_articles))
for x in my_articles:
    my_h2_tag = x.find_all("h2",{"class":"h3"})
    #print(my_h2_tag)
    for y in my_h2_tag:
        my_a_tag = y.find_all("a",{"class":"Link"})
        
        for z in my_a_tag:
            span = z.find_all("span")

        #print(type(span))
        span_strings = [tag.string for tag in span]
        clean_strings = [x.strip() for x in span_strings]
        print(clean_strings)
        for i in my_a_tag:
            a_text = i.find_all("a",{"id":"text"})
            print(a_text)
````

#### Retrieve an arbitary Wikipedia page of "Python" and create a list of links on that page.

````python
# This is the url you will scrape in this exercise
url ='https://en.wikipedia.org/wiki/Python' 
````

````python
# your code here
response = requests.get(url)
html = response.content
parsed_html = bs4.BeautifulSoup(html, 'html.parser')
my_a = parsed_html.find_all("a",{"href":True})

#print(my_a)
list_links = []

for x in my_a:
    list_links.append(x["href"])
    if not x["href"].startswith(url):
        list_links.append(url + x["href"])
print("Mi lista -->")
print(list_links)
print(len(list_links))
````

#### Find the number of titles that have changed in the United States Code since its last release point.

````python
# This is the url you will scrape in this exercise
url = 'http://uscode.house.gov/download/download.shtml'
````

````python
# your code here
response = requests.get(url)
html = response.content
parsed_html = bs4.BeautifulSoup(html,"html.parser")
my_divs = parsed_html.find_all("div",{"class":"uscitem"})
final_list = []
for x in my_divs:
    div = x.find_all("div",{"class":"usctitlechanged"})
    if div:
        final_list.append(div)
print("Final list: ", final_list)

````

````python
for element in final_list:
    
    print(type(element))
    final_string = [tag.string for tag in element ]
    final_cleaned_string = [x.strip() for x in final_string]
    print(final_cleaned_string)
print("Number of elements: ", str(len(final_list)))
````
