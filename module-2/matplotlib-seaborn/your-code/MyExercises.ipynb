```python
import matplotlib.pyplot as plt   # conda install matplotlib
```

# LINE PLOT

```python

# Pais 1
x = [2016,2017,2018,2019,2020,2021]
y = [42,43,45,47,48,50]

# Pais 2
x2 = [2016,2017,2018,2019,2020,2021]
y2 = [43,43,44,44,45,45]
```

```python
plt.plot(x,y, marker=".", linestyle=':', color="cyan", label="Hungria")
plt.plot(x2,y2, marker=".", linestyle='--', color="red", label="Honduras")

plt.xlabel("Años")
plt.ylabel("Poblacion en millones")
plt.title("Años vs Poblacion")
plt.yticks([45,48,51])# Numeros que queremos ver

#leyenda
plt.legend(loc='lower right')
plt.show()
```
