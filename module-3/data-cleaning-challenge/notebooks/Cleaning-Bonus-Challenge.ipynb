```python
# Filling nulls with median
df1["petal_width_cm"] = df1["petal_width_cm"].fillna(df1["petal_width_cm"].median())
```

```python
# Checking correlations between numeric series
num_series = df1[["sepal_length_cm", "sepal_width_cm", "petal_length_cm", "petal_width_cm"]]

num_correlation = num_series.corr()


plt.figure(figsize=(12,8))
cmap = sns.diverging_palette(230, 20, as_cmap=True)

sns.heatmap(num_correlation,cmap=cmap, annot=True)
```

```python
plt.figure(figsize=(12, 8))
df1['petal_width_cm'].plot(kind="hist", subplots=True, layout=(1,1), title='Petal Width (cm) - Antes')
plt.show()
```
